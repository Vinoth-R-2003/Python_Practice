```python
def discounted_amount(price):
    discount = price * 0.15 # (0.15 =15/100)
    final_price = price - discount
    return final_price
original_price = int(input("Enter the Total amount : "))
dis_price = discounted_amount(original_price)
print("The discounted Amount is ",dis_price)
```

```python
def getGrade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "Fail"
score=int(input("Enter the score : "))
print(" Grade ",getGrade(score))
```

```python
def calculate(price, quantity, tax=0.1, shipping_fee=50):
    subtotal = price * quantity
    tax_amount = subtotal * tax
    if subtotal >= 500:
        shipping = 0
    else:
        shipping = shipping_fee
    total = subtotal + tax_amount + shipping
    return total
price=int(input("Enter the Price :"))
quantity=int(input("Enter the quantity:"))
print(calculate(price,quantity))
print(calculate(price,quantity))
```

```python
from datetime import datetime
import math
start_str = "2026-01-01T00:00:00Z"
end_str   = "2026-07-02T12:00:00Z"
start_dt = datetime.fromisoformat(start_str.replace("Z", "+00:00"))
end_dt   = datetime.fromisoformat(end_str.replace("Z", "+00:00"))
uptime_seconds = (end_dt - start_dt).total_seconds()
SECONDS_PER_JULIAN_YEAR = 365.25 * 86400.0
year_fraction = uptime_seconds / SECONDS_PER_JULIAN_YEAR
numerator, denominator = year_fraction.as_integer_ratio()
common_factor = math.gcd(numerator, denominator)
reduced_num = numerator // common_factor
reduced_den = denominator // common_factor
print("Total Uptime (seconds):", uptime_seconds)
print("Fraction of year (decimal):", year_fraction)
print("Fraction of year (reduced ratio):", str(reduced_num) + "/" + str(reduced_den))
```
