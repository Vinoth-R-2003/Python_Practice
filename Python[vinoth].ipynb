```python
# FizzBuzz Variant(FOR)
for i in range(0,51):
  if(i%3==0 and i%5==0):
    print("Fizzbuzz")
  elif(i%3==0):
    print("Fizz")
  elif(i%5==0):
    print("Buzz")
  else:
    print(i)
  i=i+1
```

```python
# FizzBuzz Variant(While)
i=1
while(i<=50):
  if(i%3==0 and i%5==0):
    print("Fizzbuzz")
  elif(i%3==0):
    print("Fizz")
  elif(i%5==0):
    print("Buzz")
  else:
    print(i)
  i=i+1
```

```python
# Odd/Even List Separator(For) :
n=int(input("Enter the range :"))
EvenCount=0
OddCount=0
for i in range(1,n+1):
  if(i%2==0):
    print(i," is Even")
    EvenCount+=1
  else:
    print(i,(" is Odd"))
    OddCount+=1
print("TOTAL ODD NUMBERS : ",OddCount)
print("TOTAL EVEN NUMBERS : ",EvenCount)
```

```python
# Number Guessing Game (Fixed Tries):
secretNum=67
Chances=5
while(Chances>0):
  guess=int(input("Enter your guess :"))
  if(guess<secretNum):
    print("Too low")
  elif(guess>secretNum):
    print("Too high")
  else:
    print("CORRECT!!!!")
    print("You Won")
    break
  Chances-=1
  print("You Lost!!")
  print("better luck next time")
```

```python
# Running Sum with Negative Filter:
total=0
while True:
  num=int(input("Enter the number to add (0 to stop) : "))
  if(num>0):
    total+=num
  elif(num==0):
    break
  else:
    print("Negative Number!")
print("Total sum :",total)
```
